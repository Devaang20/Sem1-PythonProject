```python
import mysql.connector
import pandas as pd

# Connect to database
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="manager",
    database="pythpro"
)

# Fetch data
df = pd.read_sql("SELECT * FROM student", conn)
conn.close()

# Configure pandas to show everything
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# Print full table
print(df)
```

```python
# Number of students
print("Total Students:", len(df))

# Gender distribution
print("\nGender Distribution:")
print(df['gender'].value_counts())

# Coaching type distribution
print("\nCoaching Type Distribution:")
print(df['coaching'].value_counts())

# Class X performance
print("\nClass X Performance:")
print(df['class_X_Percentage'].value_counts())
```

```python
import matplotlib.pyplot as plt

df['Gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    figsize=(5,5),
    title="Gender Distribution"
)
plt.ylabel('')
plt.show()
```

```python
df['coaching'].value_counts().plot(
    kind='bar',
    title="Coaching Type Distribution",
    xlabel="Coaching Type",
    ylabel="Number of Students",
    figsize=(6,4)
)
plt.show()
```
